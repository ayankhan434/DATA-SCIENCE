### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def level(x):
  if x>=90:
    return 'HIGH'
  elif x>=70:
    return "Medium"
  elif x<70:
    return "low"
```

```python
df['scoring_level'] = df['total_score'].apply(level)
```

### Create a new column study_category

```python
def category(x):
  if x<10:
    return "light"
  elif x<=20:
    return "Moderaate"
  else:
    return "heavy"
```
