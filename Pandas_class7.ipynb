## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python

# Dataset 1: Product details
import pandas as pd

df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Ayan', 'Rahul', 'Priya', 'Krishna', 'Amit', 'Neha'],
    'Age': [20, 21, 19, 20, 22, 19],
    'City': ['Delhi', 'Meerut', 'Noida', 'Delhi', 'Meerut', 'Noida'],
    'Marks': [85, 72, 90, 65, 78, 95]
})

df1

# Dataset 2: More products (same columns as df1 → good for vertical stacking)
df2 = pd.DataFrame({

    'Maths': [88, 70, 95, 60, 80, 75],
    'Python': [92, 75, 89, 72, 85, 90],
    'Attendance': [90, 80, 95, 70, 85, 88],
    'Study_Hours': [5, 3, 6, 2, 4, 5]
})

df2
# Dataset 3: Additional info (different columns → good for horizontal stacking)
df3 = pd.DataFrame({

    'Course': ['Python', 'Data Science', 'AI/ML', 'Web Dev', 'Python', 'Data Science'],
    'Fees': [15000, 25000, 30000, 18000, 15000, 25000],
    'Payment_Status': ['Paid', 'Pending', 'Paid', 'Paid', 'Pending', 'Paid'],
    'Scholarship': [2000, 0, 5000, 0, 1500, 3000]
})

df3
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
pd.concat([df1,df3],axis=0)
```

```python
pd.concat([df1,df3],axis=1)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
Marks=pd.concat([df1,df2],axis=1,ignore_index=False)

Marks
```

```python

College=pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 111,],
    "location":["kanpur","bhubneshwar",'delhi',"kanpur",'mumbai','narwal']

})
College
```

```python
pd.merge(Marks,College,on="Student_ID", how='inner')
```

```python
pd.merge(Marks,College,on="Student_ID", how='left')
```

```python
pd.merge(Marks,College,on="Student_ID", how='right')
```

```python
pd.merge(Marks,College,on="Student_ID", how='outer')
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
products=pd.concat([df1,df2], ignore_index=True)
products
```

```python
products.set_index('Student_ID',inplace=True)
products
```
